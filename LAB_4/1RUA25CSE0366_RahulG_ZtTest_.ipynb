## Quick Reference

| | Z-test | t-test |
|---|---|---|
| Use when | Population σ is **known** | Population σ is **unknown** |
| Formula | $z = \dfrac{\bar{x} - \mu_0}{\sigma / \sqrt{n}}$ | $t = \dfrac{\bar{x} - \mu_0}{s / \sqrt{n}}$, $df = n - 1$ |

| Tail | $H_1$ | p-value (Z) | p-value (t) | Critical value (Z, α = 0.05) |
|---|---|---|---|---|
| Left | $\mu < \mu_0$ | `stats.norm.cdf(z)` | `stats.t.cdf(t, dof)` | −1.645 |
| Right | $\mu > \mu_0$ | `1 - stats.norm.cdf(z)` | `1 - stats.t.cdf(t, dof)` | +1.645 |
| Two | $\mu \neq \mu_0$ | `2 * (1 - stats.norm.cdf(abs(z)))` | `2 * (1 - stats.t.cdf(abs(t), dof))` | ±1.960 |

**Decision rule:** if **p-value ≤ α**, reject $H_0$. Otherwise, fail to reject $H_0$.

For the sample standard deviation use `df['column'].std()`. pandas uses `ddof=1` by default, which is what the t-test needs.

⚠️ The dataset is stored in a variable called `df`. Store the degrees of freedom in a variable called **`dof`**, not `df`, or you will overwrite the dataset.

# Part A – Z-Tests (σ known, use all 40 students)

## Q1. Z-Test – Left-Tailed

A health report says college students sleep **7 hours** per night on average, with a known population standard deviation of **σ = 1.2 hours**.
Using `sleep_hours` for **all 40 students**, test at α = 0.05 whether students sleep **less than 7 hours** on average.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q1 – YOUR CODE HERE =====================
# Z-test, left-tailed: column 'sleep_hours', mu_0 = 7, sigma = 1.2, all 40 students
# ========================= Q1 - Z-TEST, LEFT-TAILED =========================

# Given values
mu_0 = 7          # Hypothesized population mean
sigma = 1.2       # Known population standard deviation
alpha = 0.05

# Sample data
x = df['sleep_hours']
n = len(x)

# (a) Hypotheses
H0 = "μ = 7 hours"
H1 = "μ < 7 hours"

# (b) Calculate sample mean and z-statistic
x_bar = x.mean()

z = (x_bar - mu_0) / (sigma / np.sqrt(n))

# (c) Critical value for left-tailed test
critical_value = stats.norm.ppf(alpha)

# (d) p-value for left-tailed test
p_value = stats.norm.cdf(z)

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that college students sleep less than 7 hours per night on average."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that college students sleep less than 7 hours per night on average."

# Display answers
print("Your answer – Q1")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Left-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Test statistic (z):", round(z, 4))
print("Critical value:", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)
```

## Q2. Z-Test – Right-Tailed

A national survey says young adults spend **5 hours** a day on screens, with a known population standard deviation of **σ = 1.5 hours**.
Using `screen_time` for **all 40 students**, test at α = 0.05 whether these students' screen time is **more than 5 hours** a day.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q2 – YOUR CODE HERE =====================
# Z-test, right-tailed: column 'screen_time', mu_0 = 5, sigma = 1.5, all 40 students
# ========================= Q1 - Z-TEST, LEFT-TAILED =========================
# ========================= Q2 - Z-TEST, RIGHT-TAILED =========================

# Given values
mu_0 = 5
sigma = 1.5
alpha = 0.05

# Sample data
x = df['screen_time']
n = len(x)

# (a) Hypotheses
H0 = "μ = 5 hours"
H1 = "μ > 5 hours"

# (b) Find n and sample mean
x_bar = x.mean()

# (c) Calculate z-statistic
z = (x_bar - mu_0) / (sigma / np.sqrt(n))

# (d) Critical value for right-tailed test
critical_value = stats.norm.ppf(1 - alpha)

# p-value for right-tailed test
p_value = 1 - stats.norm.cdf(z)

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that these students spend more than 5 hours per day on screens."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that these students spend more than 5 hours per day on screens."

# Display answers
print("Your answer – Q2")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Right-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Test statistic (z):", round(z, 4))
print("Critical value:", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)
# Given values
```

## Q3. Z-Test – Two-Tailed

Over many years, the course's final exam has had a mean of **65** with a known standard deviation of **σ = 10**.
Using `exam_score` for **all 40 students**, test at α = 0.05 whether this year's mean score is **different from 65**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) find $n$ and $\bar{x}$, (c) compute $z$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion.

*(15 marks)*

```python
# ===================== Q3 – YOUR CODE HERE =====================
# Z-test, two-tailed: column 'exam_score', mu_0 = 65, sigma = 10, all 40 students
# ========================= Q3 - Z-TEST, TWO-TAILED =========================

# Given values
mu_0 = 65
sigma = 10
alpha = 0.05

# Sample data
x = df['exam_score']
n = len(x)

# (a) Hypotheses
H0 = "μ = 65"
H1 = "μ ≠ 65"

# (b) Find sample mean
x_bar = x.mean()

# (c) Calculate z-statistic
z = (x_bar - mu_0) / (sigma / np.sqrt(n))

# (d) Critical value for two-tailed test
critical_value = stats.norm.ppf(1 - alpha / 2)

# p-value for two-tailed test
p_value = 2 * (1 - stats.norm.cdf(abs(z)))

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that the mean exam score is different from 65."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that the mean exam score is different from 65."

# Display answers
print("Your answer – Q3")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Two-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Test statistic (z):", round(z, 4))
print("Critical value: ±", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)
```

## Q4. t-Test – Left-Tailed

The class teacher of **Section B** believes her students' attendance is **below 83 %**. The population standard deviation is **not known**.
Using `attendance` for **Section B only**, test at α = 0.05 whether the mean attendance is **less than 83 %**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q4 – YOUR CODE HERE =====================
# t-test, left-tailed: column 'attendance', Section B only, mu_0 = 83, sigma unknown
# ========================= Q4 - T-TEST, LEFT-TAILED =========================

# Given values
mu_0 = 83
alpha = 0.05

# (a) Hypotheses
H0 = "μ = 83%"
H1 = "μ < 83%"

# Select Section B only
section_B = df[df['section'] == 'B']['attendance']

# (b) Find n, x̄ and df
n = len(section_B)
x_bar = section_B.mean()
s = section_B.std(ddof=1)
df_value = n - 1

# (c) Calculate t-statistic
t = (x_bar - mu_0) / (s / np.sqrt(n))

# (d) Critical value and p-value
critical_value = stats.t.ppf(alpha, df_value)
p_value = stats.t.cdf(t, df_value)

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that the mean attendance is below 83%."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that the mean attendance is below 83%."

# (f) Verification using scipy
t_stat, p_val = stats.ttest_1samp(
    section_B,
    popmean=mu_0,
    alternative='less'
)

# Display results
print("Your answer – Q4")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Left-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Sample standard deviation (s):", round(s, 4))
print("Degrees of freedom (df):", df_value)
print("Test statistic (t):", round(t, 4))
print("Critical value:", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)

print("\nVerification using scipy.stats.ttest_1samp:")
print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_val, 6))
```

## Q5. t-Test – Right-Tailed

The college recommends at least **2.5 hours** of self-study per day. The teacher of **Section A** claims her students study **more than 2.5 hours** a day on average. σ is **not known**.
Using `study_hours` for **Section A only**, test the claim at α = 0.05.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q5 – YOUR CODE HERE =====================
# t-test, right-tailed: column 'study_hours', Section A only, mu_0 = 2.5, sigma unknown
# ========================= Q5 - T-TEST, RIGHT-TAILED =========================

# Given values
mu_0 = 2.5
alpha = 0.05

# (a) Hypotheses
H0 = "μ = 2.5 hours"
H1 = "μ > 2.5 hours"

# (b) Select Section A only
section_A = df[df['section'] == 'A']['study_hours']

# Find n, sample mean, sample standard deviation and df
n = len(section_A)
x_bar = section_A.mean()
s = section_A.std(ddof=1)
df_value = n - 1

# (c) Calculate t-statistic
t = (x_bar - mu_0) / (s / np.sqrt(n))

# (d) Critical value and p-value
critical_value = stats.t.ppf(1 - alpha, df_value)
p_value = 1 - stats.t.cdf(t, df_value)

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that Section A students study more than 2.5 hours per day on average."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that Section A students study more than 2.5 hours per day on average."

# (f) Verify using scipy
t_stat, p_val = stats.ttest_1samp(
    section_A,
    popmean=mu_0,
    alternative='greater'
)

# Display results
print("Your answer – Q5")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Right-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Sample standard deviation (s):", round(s, 4))
print("Degrees of freedom (df):", df_value)
print("Test statistic (t):", round(t, 4))
print("Critical value:", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)

print("\nVerification using scipy.stats.ttest_1samp:")
print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_val, 6))
```

## Q6. t-Test – Two-Tailed

The transport office plans the college bus timetable assuming an average one-way commute of **35 minutes**. σ is **not known**.
Using `commute_min` for **Section C only**, test at α = 0.05 whether the mean commute time is **different from 35 minutes**.

**Steps:** (a) write $H_0$, $H_1$ and the tail, (b) select the rows for the section, then find $n$, $\bar{x}$, $s$ and $df$, (c) compute $t$, (d) find the critical value and the p-value, (e) make a decision and write a conclusion, (f) check your answer with `stats.ttest_1samp(data, popmean, alternative=...)`.

*(15 marks)*

```python
# ===================== Q6 – YOUR CODE HERE =====================
# t-test, two-tailed: column 'commute_min', Section C only, mu_0 = 35, sigma unknown
# ========================= Q6 - T-TEST, TWO-TAILED =========================

# Given values
mu_0 = 35
alpha = 0.05

# (a) Hypotheses
H0 = "μ = 35 minutes"
H1 = "μ ≠ 35 minutes"

# (b) Select Section C only
section_C = df[df['section'] == 'C']['commute_min']

# Find n, sample mean, sample standard deviation and df
n = len(section_C)
x_bar = section_C.mean()
s = section_C.std(ddof=1)
df_value = n - 1

# (c) Calculate t-statistic
t = (x_bar - mu_0) / (s / np.sqrt(n))

# (d) Critical value and p-value
critical_value = stats.t.ppf(1 - alpha / 2, df_value)

# Two-tailed p-value
p_value = 2 * (1 - stats.t.cdf(abs(t), df_value))

# (e) Decision
if p_value < alpha:
    decision = "Reject H₀"
    conclusion = "There is sufficient evidence to conclude that the mean commute time is different from 35 minutes."
else:
    decision = "Fail to reject H₀"
    conclusion = "There is insufficient evidence to conclude that the mean commute time is different from 35 minutes."

# (f) Verify using scipy
t_stat, p_val = stats.ttest_1samp(
    section_C,
    popmean=mu_0,
    alternative='two-sided'
)

# Display results
print("Your answer – Q6")
print("-----------------------------------")
print("H₀:", H0)
print("H₁:", H1)
print("Tail: Two-tailed")
print("Sample size (n):", n)
print("Sample mean (x̄):", round(x_bar, 4))
print("Sample standard deviation (s):", round(s, 4))
print("Degrees of freedom (df):", df_value)
print("Test statistic (t):", round(t, 4))
print("Critical value: ±", round(critical_value, 4))
print("p-value:", round(p_value, 6))
print("Decision:", decision)
print("Conclusion:", conclusion)

print("\nVerification using scipy.stats.ttest_1samp:")
print("t-statistic:", round(t_stat, 4))
print("p-value:", round(p_val, 6))
```
