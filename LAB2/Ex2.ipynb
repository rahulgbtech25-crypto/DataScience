Data wrangling and transformation is the process of converting raw student placement data into a clean, consistent, and machine-learning-ready dataset. In this experiment, the given student placement dataset is first loaded into a Pandas DataFrame and explored by examining its rows, columns, data types, and descriptive statistics. ## Step 6  Option 1: Identifying Outliers Using Z-Scores
The value of 3 in the context of Z-scores is often used as a threshold to identify outliers in a dataset. A Z-score represents how many standard deviations a data point is away from the mean of the dataset. Specifically:

A Z-score of 0 means the data point is exactly at the mean.
A Z-score of 1 means the data point is one standard deviation above the mean, and so on.
A Z-score of 3 corresponds to a data point being 3 standard deviations away from the mean. For a normal distribution, about 99.7% of the data points fall within 3 standard deviations of the mean (according to the 68-95-99.7 rule, which describes the spread of data in a normal distribution). Therefore, points with Z-scores greater than 3 or less than -3 are considered unusually far from the mean and are often flagged as outliers.

This threshold (Z > 3 or Z < -3) is commonly used in many statistical applications because it captures the extreme values that are rare in a normal distribution, which are typically considered to be outliers. However, the choice of threshold can vary depending on the specific application and the nature of the data.

```python
columns_to_check=["salary"]
z_score=stats.zscore(df[columns_to_check])
u=(z_score>3)
l=(z_score<3)
indx = u | l
clean_df=df[~indx]
clean_df.info()
```

```python
l1=df["salary"].quantile(0.05)
u1=df["salary"].quantile(0.95)
df_capped=df.copy()
df_capped["salary"]=df_capped["salary"].clip(11,u1)
df_capped.info()
df.head()
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv("Automobile.csv")

# Display first 10 rows
display(df.head(10))

# Display shape
print("Shape:", df.shape)
```

```python
# Step 2: Explore the Dataset

# Column names
print("Column names:")
display(df.columns)

# Data types
print("Data types:")
display(df.dtypes)

# Descriptive statistics
print("Descriptive statistics:")
display(df.describe())

# Numerical columns
numerical_columns = df.select_dtypes(include=np.number).columns
print("Numerical columns:")
display(numerical_columns)

# Categorical columns
categorical_columns = df.select_dtypes(include=["object", "string", "category"]).columns
print("Categorical columns:")
display(categorical_columns)

# Unique values in categorical columns
for column in categorical_columns:
    print("\n", column)
    print(df[column].unique())
```

```python
# Step 3: Check Missing Values

# Number of missing values
print("Number of missing values:")
display(df.isnull().sum())

# Percentage of missing values
print("Percentage of missing values:")
display(df.isnull().mean().mul(100).round(2))
```

```python
# Step 4: Handle Missing Values

# Replace ? in horsepower with NaN first
df["horsepower"] = df["horsepower"].replace("?", np.nan)

# Convert horsepower to numeric
df["horsepower"] = pd.to_numeric(df["horsepower"])

# Numerical columns
numerical_columns = [
    "cylinders",
    "displacement",
    "horsepower",
    "weight",
    "acceleration"
]

# Fill numerical missing values with median
df[numerical_columns] = df[numerical_columns].fillna(
    df[numerical_columns].median()
)

# Fill categorical missing values with mode
if df["origin"].isnull().sum() > 0:
    df["origin"] = df["origin"].fillna(df["origin"].mode()[0])

# Verify
print("Missing values after handling:")
display(df.isnull().sum())
```

```python
# Step 5: Remove Duplicate Records

# Check duplicate rows
print("Number of duplicate records:")
print(df.duplicated().sum())

# Display duplicate rows
display(df[df.duplicated()])

# Remove duplicates
df = df.drop_duplicates()

# Verify
print("Duplicates after removal:")
print(df.duplicated().sum())

print("New shape:", df.shape)
```

```python
# Step 6: Clean the horsepower Column

# Identify non-numeric values
print("Non-numeric horsepower values:")
display(df[df["horsepower"] == "?"])

# Replace ? with NaN
df["horsepower"] = df["horsepower"].replace("?", np.nan)

# Convert to numeric
df["horsepower"] = pd.to_numeric(df["horsepower"])

# Handle missing horsepower values using mean
df["horsepower"] = df["horsepower"].fillna(
    df["horsepower"].mean()
)

# Verify
print("Missing horsepower values:")
print(df["horsepower"].isna().sum())
```

```python
# Step 7: Transform the origin Column

# Display original unique values
print("Original origin values:")
display(df["origin"].unique())

# Convert origin into meaningful labels
df["origin"] = df["origin"].replace({
    1: "usa",
    2: "europe",
    3: "japan",
    "1": "usa",
    "2": "europe",
    "3": "japan"
})

# Display transformed values
print("Transformed origin values:")
display(df["origin"].unique())
```

```python
# Step 8: Create weight_kg

df["weight_kg"] = df["weight"] * 0.453592

display(
    df[["origin", "weight", "weight_kg"]].head()
)
```

```python
# Step 9: Create mpg_category

df["mpg_category"] = np.select(
    [
        df["mpg"] < 20,
        df["mpg"].between(20, 29)
    ],
    [
        "Low",
        "Medium"
    ],
    default="High"
)

display(df[["mpg", "mpg_category"]].head(10))
```

```python
# Step 10: Create vehicle_age

df["vehicle_age"] = 2026 - df["model_year"]

display(
    df[["model_year", "vehicle_age"]].head()
)
```

```python
# Step 11: Rename Columns

df = df.rename(columns={
    "mpg": "miles_per_gallon",
    "horsepower": "hp",
    "weight": "weight_lbs",
    "model_year": "year"
})

display(df.head())
```

```python
# Step 12: Filter the Data

print("Vehicles with MPG > 30:")
display(df[df["miles_per_gallon"] > 30])

print("Vehicles with horsepower > 150:")
display(df[df["hp"] > 150])

print("Vehicles with cylinders >= 6:")
display(df[df["cylinders"] >= 6])

print("Vehicles manufactured after 1980:")
display(df[df["year"] > 1980])

print("Vehicles originating from USA:")
display(df[df["origin"] == "usa"])
```

```python
# Step 13: Label Encoding

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

# Apply Label Encoding
df["origin_encoded"] = le.fit_transform(df["origin"])

# Display original and encoded values
display(
    df[["origin", "origin_encoded"]].drop_duplicates()
)

# Display category-to-label mapping
mapping = dict(
    zip(
        le.classes_,
        le.transform(le.classes_)
    )
)

print("Category-to-label mapping:")
display(mapping)
```

```python
# Step 14: One-Hot Encoding

# Create one-hot encoded columns
origin_onehot = pd.get_dummies(
    df["origin"],
    prefix="origin"
)

# Display one-hot encoded values
display(origin_onehot.head())

# Add one-hot columns to dataframe
df = pd.concat([df, origin_onehot], axis=1)

print("One-Hot Encoding is more appropriate for origin because origin is nominal categorical data and its categories have no meaningful order.")
```

```python
# Step 15: Identify Outliers Using Z-Scores

from scipy.stats import zscore

outlier_columns = [
    "cylinders",
    "displacement",
    "hp",
    "weight_lbs",
    "acceleration",
    "miles_per_gallon",
    "weight_kg",
    "vehicle_age"
]

# Calculate Z-scores
z_scores = df[outlier_columns].apply(zscore)

# Identify outliers
outliers = abs(z_scores) > 3

# Count outliers in each column
print("Number of outliers in each column:")
display(outliers.sum())

# Display rows containing outliers
print("Rows containing outliers:")
display(df[outliers.any(axis=1)])
```

```python
# Step 16: Standardization Using StandardScaler

from sklearn.preprocessing import StandardScaler

scaling_columns = [
    "cylinders",
    "displacement",
    "hp",
    "weight_lbs",
    "acceleration",
    "miles_per_gallon",
    "weight_kg",
    "vehicle_age"
]

standard_scaler = StandardScaler()
df_standardized = df.copy()

df_standardized[scaling_columns] = standard_scaler.fit_transform(
    df[scaling_columns]
)

# Display standardized values
display(df_standardized[scaling_columns].head())

# Verify mean
print("Mean:")
display(df_standardized[scaling_columns].mean())

# Verify standard deviation
print("Standard deviation:")
display(df_standardized[scaling_columns].std())
```

```python
# Step 17: Normalization Using MinMaxScaler

from sklearn.preprocessing import MinMaxScaler

minmax_scaler = MinMaxScaler()

df_normalized = df.copy()

df_normalized[scaling_columns] = minmax_scaler.fit_transform(
    df[scaling_columns]
)

# Display normalized values
display(df_normalized[scaling_columns].head())

# Verify minimum
print("Minimum values:")
display(df_normalized[scaling_columns].min())

# Verify maximum
print("Maximum values:")
display(df_normalized[scaling_columns].max())
```

```python
# Step 18: Create X and Y Variables

# Create X by removing the target and target-derived category
X = df.drop(
    columns=[
        "miles_per_gallon",
        "mpg_category",
        "origin"
    ]
)

# Create Y as the target variable
Y = df["miles_per_gallon"]

# Display shapes
print("Shape of X:", X.shape)
print("Shape of Y:", Y.shape)

# Combine X and Y
XY = pd.concat([X, Y], axis=1)

# Save X and Y
XY.to_csv(
    "automobile_X_Y.csv",
    index=False
)

print("automobile_X_Y.csv saved successfully.")

# Load CSV again
loaded_data = pd.read_csv(
    "automobile_X_Y.csv"
)

# Display first 5 rows
display(loaded_data.head())
```

```python
# Step 19: Save the Preprocessed Dataset

# Create final dataset using standardized numerical features
final_df = df_standardized.drop(
    columns=[
        "origin",
        "mpg_category"
    ]
)

# Display final dataset
display(final_df)

# Check missing values
print("Missing values:")
display(final_df.isnull().sum())

# Save final dataset
final_df.to_csv(
    "automobile_preprocessed.csv",
    index=False
)

print("automobile_preprocessed.csv saved successfully!")
```
